```python
import pandas as pd


data = {
    'Roll No': [1, 1, 2, 2, 1, 2],
    'Subject': ['DS', 'Java', 'DS', 'Java', 'DS', 'Java'],
    'Marks': [25, 40, 30, 26, 40, 50],
    'Date': ['01-01-23', '02-01-23', '01-01-23', '02-01-23', '15-01-23', '15-01-23']
}
df = pd.DataFrame(data)
df
```

```python

df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%y')
df = df.sort_values(by=['Roll No', 'Subject', 'Date'], ascending=[True, True, False])
```

```python
df['attempt_rank'] = df.groupby(['Roll No', 'Subject']).cumcount()+1
df_top3 = df[df['attempt_rank'] <=3].copy()
```

```python

marks_df = df_top3.pivot(index=['Roll No', 'Subject'], columns='attempt_rank', values='Marks').fillna(0).reindex(columns=[1,2,3], fill_value=0)
marks_df.columns  =['M1', 'M2', 'M3']
marks_df = marks_df.astype(int).reset_index()
```

```python

latest_date = df_top3[df_top3['attempt_rank'] == 1][['Roll No', 'Subject', 'Date']]
```

```python
 
new_df = pd.merge(marks_df, latest_date, on=['Roll No', 'Subject'])
new_df['Date'] = new_df['Date'].dt.strftime('%d-%m-%y')
print(new_df)
```
