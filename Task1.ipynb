```python
import pandas as pd

df=pd.DataFrame(
    {'Rollno':[1,1,2,2,1,2],
    'Subject':['DS','Java','DS','Java','DS','Java'],
    'Marks':[25,40,30,26,40,50],
    'Date':['01-01-23','02-02-23','01-01-23','02-01-23','15-01-23','15-01-23']}
    )
```

```python
df
```

```python
df['Date']=pd.to_datetime(df['Date'],format='%d-%m-%y')
df['Date']
```

```python
df=df.sort_values('Date',ascending=False)
df
```

```python
g=df.groupby(['Rollno','Subject'])
```

```python
marks=g['Marks'].apply(lambda x:(list(x.head(2))+[0,0,0])[:3])
outputdf=pd.DataFrame(marks.tolist(),index=marks.index,columns=['M1','M2','M3'])
outputdf
```

```python
outputdf['Date']=g['Date'].max().dt.strftime('%d-%m-%y')

outputdf=outputdf.reset_index()
print(outputdf)
```
